```python
from diffusers import DiffusionPipeline
import torch
```

```python
model_id = "paceailab/StableDiffusion"
pipe = DiffusionPipeline.from_pretrained("paceailab/StableDiffusion", torch_dtype=torch.float16, use_safetensors=True, variant="fp16")
```

```python
prompt = "Beautiful African Wax Pattern with orange and black design"
```

```python
image = pipe(prompt=prompt).images[0]
```

```python
image.save("african_wax_pattern.png")
```
